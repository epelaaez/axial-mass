# Z-expansion truncation

Compute the range for $z$ given $t_\text{cut}, t_0, k_\text{max}$, and a $Q^2$ range. 

```python
import numpy as np

def z(Q2, t_cut, t0):
    """Conformal map from Q^2 to z."""
    return ((np.sqrt(t_cut + Q2) - np.sqrt(t_cut - t0)) /
            (np.sqrt(t_cut + Q2) + np.sqrt(t_cut - t0)))

# (label, t_cut [GeV^2], t0 [GeV^2], k_max)
configs = [
    ("MINERvA 2023",      0.175318, -0.75, 8),
    ("MINERvA 2026",      0.161604, -0.50, 7),
    ("MINERvA 2026",      0.161604, -0.50, 6),
    ("LQCD 2026",         0.161604, -0.50, 6),
    ("MINERvA+LQCD 2026", 0.161604, -0.50, 6),
]

Q2_min, Q2_max = 0.0, 2.0   # fit range in GeV^2

for label, t_cut, t0, k_max in configs:
    z_lo = z(Q2_min, t_cut, t0)
    z_hi = z(Q2_max, t_cut, t0)
    z_abs = max(abs(z_lo), abs(z_hi))
    trunc = z_abs ** (k_max + 1)   # size of first neglected term
    print(f"{label:20s} k_max={k_max}  "
          f"z in [{z_lo:+.3f}, {z_hi:+.3f}]  "
          f"|z|max={z_abs:.3f}  |z|^(k_max+1)={trunc:.1e}")
```
